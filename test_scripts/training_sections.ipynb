# Sectioned training: SNN calorimeter

Runs the full pipeline (generate -> train -> evaluate -> rename figures) for
five detector regions, used in the radial sectioning study:

| Section   | Spatial selection            |
|-----------|------------------------------|
| Primaries | primary cubelets only        |
| Core      | radius 0.0 - 2.5             |
| Ring      | radius 2.5 - 4.5             |
| Corona    | radius 4.5 - 15.0           |
| Total     | radius 0.0 - 15.0 (whole)   |

Every section evaluates with `--analyze`, so the deep diagnostic figures are
produced. Output figures are renamed with a per-section suffix to avoid
overwriting between runs.

**Note:** `!ren` is the Windows rename command. On Linux/macOS replace each
`!ren a.png b.png` with `!mv a.png b.png`.


## 1. Primaries

Primary cubelets only (the `PrimaryOnly/Uniform` dataset).

```python
# 1. Generate dataset (primary cubelets only)
!python -u generate_dataset.py --data-dir ../Data/PrimaryOnly/Uniform --out Dataset_Primaries.pt --target Epos --primary-only
```

```python
# 2. Train model
!python -u train_model.py --cache Dataset_Primaries.pt --epochs 10 --lr 5e-4 --model-out Model_Primaries.pth

# 3. Evaluate with deep diagnostics
!python -u print_predictions.py --model Model_Primaries.pth --cache Dataset_Primaries.pt --analyze

# Rename output figures with the 'Primaries' suffix
!ren loss.png loss_Primaries.png
!ren pred_vs_target.png pred_vs_target_Primaries.png
!ren residuals.png residuals_Primaries.png
!ren diagnostic_profile.png diagnostic_profile_Primaries.png
!ren diagnostic_bias_var.png diagnostic_bias_var_Primaries.png
!ren scatter_target_vs_predict_energy.png scatter_target_vs_predict_energy_Primaries.png
!ren diagnostic_cdf_comparison.png diagnostic_cdf_comparison_Primaries.png
!ren diagnostic_population_variance.png diagnostic_population_variance_Primaries.png
!ren diagnostic_neuron_activity.png diagnostic_neuron_activity_Primaries.png
```
