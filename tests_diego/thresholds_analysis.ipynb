```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Load data
# -----------------------------
path = "/lhome/ext/uovi123/uovi123k/SNN-Cal/test_scripts/learned_threshold_exponents_BUENO_e5_lr-3.txt"

# 4 columns: threshold exponents in log10-space
df = pd.read_csv(
    path,
    sep=r"\s+",
    header=None,
    names=["thr0", "thr1", "thr2", "thr3"]
)

print("Shape:", df.shape)
display(df.head())

# -----------------------------
# Basic stats
# -----------------------------
stats = df.describe().T
stats["median"] = df.median()
stats["std"] = df.std()
stats["var"] = df.var()
stats["iqr"] = df.quantile(0.75) - df.quantile(0.25)
stats["min"] = df.min()
stats["max"] = df.max()
display(stats)

# -----------------------------
# Extra derived quantities
# -----------------------------
# Spacing in log10-space
df["d01"] = df["thr1"] - df["thr0"]
df["d12"] = df["thr2"] - df["thr1"]
df["d23"] = df["thr3"] - df["thr2"]
df["span"] = df["thr3"] - df["thr0"]

# Linear thresholds
for i in range(4):
    df[f"lin_thr{i}"] = 10 ** df[f"thr{i}"]

spacing_stats = df[["d01", "d12", "d23", "span"]].describe().T
spacing_stats["median"] = df[["d01", "d12", "d23", "span"]].median()
display(spacing_stats)

# Check ordering
ordered = ((df["thr0"] < df["thr1"]) &
           (df["thr1"] < df["thr2"]) &
           (df["thr2"] < df["thr3"]))
print("All rows ordered:", ordered.all())
print("Ordered rows:", ordered.sum(), "/", len(ordered))

# -----------------------------
# Histograms per column
# -----------------------------
fig, axes = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)

cols = ["thr0", "thr1", "thr2", "thr3"]
for ax, col in zip(axes.ravel(), cols):
    ax.hist(df[col], bins=40, edgecolor="black", alpha=0.8)
    ax.set_title(f"Histogram of {col}")
    ax.set_xlabel("log10 threshold")
    ax.set_ylabel("Count")
    ax.grid(alpha=0.3)

plt.show()

# -----------------------------
# Histograms per column in linear scale with log-x axis
# -----------------------------
fig, axes = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)

lin_cols = ["lin_thr0", "lin_thr1", "lin_thr2", "lin_thr3"]
for ax, col in zip(axes.ravel(), lin_cols):
    ax.hist(df[col], bins=40, edgecolor="black", alpha=0.8)
    #ax.set_xscale("log")
    ax.set_title(f"Histogram of {col} (log-x)")
    ax.set_xlabel("threshold (linear scale)")
    ax.set_ylabel("Count")
    ax.grid(alpha=0.3)

plt.show()

# -----------------------------
# Boxplots of columns
# -----------------------------
plt.figure(figsize=(8, 5))
plt.boxplot([df[c] for c in cols], labels=cols, showfliers=True)
plt.ylabel("log10 threshold")
plt.title("Boxplots of threshold exponents")
plt.grid(alpha=0.3)
plt.show()

# -----------------------------
# Histograms of spacings
# -----------------------------
fig, axes = plt.subplots(1, 3, figsize=(14, 4), constrained_layout=True)

for ax, col in zip(axes, ["d01", "d12", "d23"]):
    ax.hist(df[col], bins=40, edgecolor="black", alpha=0.8)
    ax.set_title(f"Histogram of {col}")
    ax.set_xlabel("spacing in log10-space")
    ax.set_ylabel("Count")
    ax.grid(alpha=0.3)

plt.show()

# -----------------------------
# Scatter plots between columns
# -----------------------------
fig, axes = plt.subplots(2, 3, figsize=(15, 9), constrained_layout=True)

pairs = [("thr0", "thr1"), ("thr1", "thr2"), ("thr2", "thr3"),
         ("thr0", "thr2"), ("thr1", "thr3"), ("thr0", "thr3")]

for ax, (x, y) in zip(axes.ravel(), pairs):
    ax.scatter(df[x], df[y], s=8, alpha=0.5)
    mn = min(df[x].min(), df[y].min())
    mx = max(df[x].max(), df[y].max())
    ax.plot([mn, mx], [mn, mx], linestyle="--", linewidth=1)
    ax.set_xlabel(x)
    ax.set_ylabel(y)
    ax.set_title(f"{y} vs {x}")
    ax.grid(alpha=0.3)

plt.show()

# -----------------------------
# Per-row curves (subset + all)
# -----------------------------
plt.figure(figsize=(10, 6))
n_show = min(80, len(df))
for i in range(n_show):
    plt.plot([0, 1, 2, 3], df.loc[i, cols].values, alpha=0.25)
plt.xticks([0, 1, 2, 3], cols)
plt.ylabel("log10 threshold")
plt.title(f"Per-cubelet threshold curves (first {n_show} rows)")
plt.grid(alpha=0.3)
plt.show()

# Mean ± std curve
means = df[cols].mean()
stds = df[cols].std()

plt.figure(figsize=(7, 5))
x = np.arange(4)
plt.plot(x, means.values, marker="o", label="mean")
plt.fill_between(x, means.values - stds.values, means.values + stds.values, alpha=0.25, label="±1 std")
plt.xticks(x, cols)
plt.ylabel("log10 threshold")
plt.title("Mean and spread across channels")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# -----------------------------
# Linear-scale summary
# -----------------------------
lin_stats = df[[f"lin_thr{i}" for i in range(4)]].describe().T
lin_stats["median"] = df[[f"lin_thr{i}" for i in range(4)]].median()
display(lin_stats)

# -----------------------------
# Correlation matrix
# -----------------------------
plt.figure(figsize=(8, 6))
corr = df[cols + ["d01", "d12", "d23", "span"]].corr()
plt.imshow(corr, interpolation="nearest")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.index)), corr.index)
plt.title("Correlation matrix")
plt.tight_layout()
plt.show()

display(corr)

# -----------------------------
# Outliers by total span
# -----------------------------
print("Top 10 widest spans:")
display(df.nlargest(10, "span")[cols + ["d01", "d12", "d23", "span"]])

print("Top 10 narrowest spans:")
display(df.nsmallest(10, "span")[cols + ["d01", "d12", "d23", "span"]])

# -----------------------------
# Optional: save processed table
# -----------------------------
# df.to_csv("threshold_analysis_table.csv", index=False)
```
