IMAGENES

```python
import numpy as np
import matplotlib.pyplot as plt

# Creamos la cuadrícula 10x10 simulando los cubelets
x, y = np.meshgrid(np.arange(10), np.arange(10))

# Calculamos la distancia al centro (4.5, 4.5) igual que en tu código
radios = np.sqrt((x - 4.5)**2 + (y - 4.5)**2)

# Definimos las 3 máscaras (nuestros filtros R1, R2 y R3)
mask_r1 = (radios >= 0.0) & (radios < 2.5)
mask_r2 = (radios >= 2.5) & (radios < 4.5)
mask_r3 = (radios >= 4.5) & (radios < 15.0)

# Preparamos la figura con 3 subgráficos
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

# Plot Región 1
axs[0].imshow(mask_r1, cmap='Reds', origin='lower')
axs[0].set_title("Región 1: Núcleo (R < 2.5)")

# Plot Región 2
axs[1].imshow(mask_r2, cmap='Oranges', origin='lower')
axs[1].set_title("Región 2: Corona Media (2.5 <= R < 4.5)")

# Plot Región 3
axs[2].imshow(mask_r3, cmap='Blues', origin='lower')
axs[2].set_title("Región 3: Corona Externa (R >= 4.5)")

# Añadimos la cuadrícula para ver los cubelets individuales
for ax in axs:
    ax.set_xticks(np.arange(-0.5, 10, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, 10, 1), minor=True)
    ax.grid(which='minor', color='black', linestyle='-', linewidth=1)
    ax.set_xticks([]) # Ocultamos los números de los ejes para que quede más limpio
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

Entrenamiento por secciones sin umbral de energía

```python
# Creamos la cuadrícula 10x10 simulando los cubelets
x, y = np.meshgrid(np.arange(10), np.arange(10))
radios = np.sqrt((x - 4.5)**2 + (y - 4.5)**2)

# Máscaras de las 3 secciones
mask_s1 = (radios >= 0.0) & (radios < 2.0)
mask_s2 = (radios >= 2.0) & (radios < 4.0)
mask_s3 = (radios >= 4.0) & (radios < 15.0)

fig, axs = plt.subplots(1, 3, figsize=(15, 5))
titulos = ["S1: Núcleo (R < 2.0)", "S2: Anillo Medio (2.0 <= R < 4.0)", "S3: Periferia (R >= 4.0)"]
mascaras = [mask_s1, mask_s2, mask_s3]
colores = ['Reds', 'Oranges', 'Blues']

for i in range(3):
    axs[i].imshow(mascaras[i], cmap=colores[i], origin='lower')
    axs[i].set_title(titulos[i], fontsize=14)
    axs[i].set_xticks(np.arange(-0.5, 10, 1), minor=True)
    axs[i].set_yticks(np.arange(-0.5, 10, 1), minor=True)
    axs[i].grid(which='minor', color='black', linestyle='-', linewidth=1)
    axs[i].set_xticks([]); axs[i].set_yticks([])

plt.tight_layout()
plt.show()
```
